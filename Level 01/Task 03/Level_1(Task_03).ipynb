```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv('Dataset.csv')
```

```python
df['sales'] = pd.to_numeric(df['sales'].astype(str).str.replace(',', ''), errors='coerce')
df['profit'] = pd.to_numeric(df['profit'], errors='coerce')
```

```python
# Date fields convert to datetime format
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
df['ship_date'] = pd.to_datetime(df['ship_date'], errors='coerce')

df['order_year'] = df['order_date'].dt.year
df['order_month'] = df['order_date'].dt.month
df['order_day'] = df['order_date'].dt.day
df['order_dayofweek'] = df['order_date'].dt.dayofweek

df['shipping_days'] = (df['ship_date'] - df['order_date']).dt.days

print("Date features extracted successfully!")
df[['order_date', 'order_year', 'order_month', 'shipping_days']].head()
```

```python
df['unit_price'] = df['sales'] / df['quantity']
df['profit_margin'] = df['profit'] / df['sales']
df.fillna(0, inplace=True)
```

```python
# Categorical Columns
categorical_cols = ['ship_mode', 'segment', 'market', 'region', 'category', 'sub_category', 'order_priority']

label_encoder = LabelEncoder()

for col in categorical_cols:
    df[col + '_encoded'] = label_encoder.fit_transform(df[col].astype(str))

print("Categorical encoding completed!")
```

```python
target_variable='profit'

drop_columns=['order_id', 'order_date', 'ship_date', 'customer_name', 'state', 'country','product_id', 'product_name'] +categorical_cols
ml_df = df.drop(columns=drop_columns)

print("Final ML-Ready Dataset Shape:", ml_df.shape)
print("Selected Target Variable:", target_variable)
ml_df.head()
```

```python
ml_df.to_csv('ml_ready_dataset.csv', index=False)

from google.colab import files
files.download('ml_ready_dataset.csv')
```
