```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
```

```python
df=pd.read_csv('Dataset.csv')
```

```python
df['sales'] = pd.to_numeric(df['sales'].astype(str).str.replace(',', ''), errors='coerce')
df['profit'] = pd.to_numeric(df['profit'], errors='coerce')
df.fillna(0, inplace=True)
```

```python
df['is_profitable'] = np.where(df['profit'] > 0, 1, 0)

cat_cols = ['ship_mode', 'segment', 'market', 'region', 'category', 'sub_category', 'order_priority']
le = LabelEncoder()
for col in cat_cols:
    df[col] = le.fit_transform(df[col].astype(str))

X_class = df.drop(columns=['order_id', 'order_date', 'ship_date', 'customer_name', 'state', 'country', 'product_id', 'product_name', 'profit', 'is_profitable'])
y_class = df['is_profitable']

clf_model = RandomForestClassifier(n_estimators=50, random_state=42)
clf_model.fit(X_class, y_class)

X_reg = df[['sales', 'quantity', 'discount', 'category', 'region', 'segment']]
y_reg = df['profit']

reg_model = RandomForestRegressor(n_estimators=50, random_state=42)
reg_model.fit(X_reg, y_reg)

print("AI Models trained successfully and ready for business predictions!")
```

```python
new_orders = pd.DataFrame({
    'ship_mode': [0, 1, 2],
    'segment': [1, 0, 2],
    'market': [0, 2, 1],
    'region': [1, 3, 0],
    'category': [0, 1, 2],
    'sub_category': [2, 5, 1],
    'sales': [450.0, 120.0, 850.0],
    'quantity': [3, 2, 5],
    'discount': [0.0, 0.4, 0.1],
    'shipping_cost': [25.0, 10.0, 50.0],
    'order_priority': [1, 2, 0],
    'year': [2024, 2024, 2024]
})

new_orders['predicted_profitability'] = clf_model.predict(new_orders)

reg_features = new_orders[['sales', 'quantity', 'discount', 'category', 'region', 'segment']]
new_orders['predicted_profit_amount'] = reg_model.predict(reg_features)

print("AI Prediction Results for New Orders:\n")
print(new_orders[['sales', 'quantity', 'discount', 'predicted_profitability', 'predicted_profit_amount']])
```

```python
profit_summary = df.groupby('category').agg({
    'profit': ['count', 'sum', 'mean'],
    'sales': 'sum'
}).reset_index()

profit_summary.columns = ['Category', 'Order_Count', 'Total_Profit', 'Avg_Profit', 'Total_Sales']
print("Category-wise Business Performance:\n")
print(profit_summary)
```
